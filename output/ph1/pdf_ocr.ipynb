```python
from pdf2image import convert_from_path
import pytesseract
```

```python
pdf_file = "Preface.pdf"
```

```python
# Convert PDF to a list of PIL images
pages = convert_from_path(pdf_file)

ocr_result = []

# Perform OCR on each page with Arabic language specified
for i, page in enumerate(pages):
    text = pytesseract.image_to_string(page, lang='ara') # Specify Arabic language
    ocr_result.append(f'--- Page {i+1} (Arabic) ---\n{text}')

# Combine all OCR results into a single string
arabic_ocr_result = '\n\n'.join(ocr_result)
print(arabic_ocr_result)
```

```python
# Save the extracted Arabic text to a file
output_filename = 'extracted_arabic_text.txt'
with open(output_filename, 'w', encoding='utf-8') as f:
    f.write(arabic_ocr_result)

print(f'Extracted Arabic text saved to {output_filename}')
```
