## Approach 1

## Approach 2

```python
from pdf2image import convert_from_path
import pytesseract

# Path to your PDF file
pdf_path = 'page_5.pdf'

# 1. Convert PDF to a list of PIL Image objects
# Setting 300 DPI is recommended for better OCR accuracy
pages = convert_from_path(pdf_path, 300)

ocr_result_2 = ""

# 2. Iterate through pages and extract text
for page in pages:
    text = pytesseract.image_to_string(page, lang="ara")
    ocr_result_2 += text + "\n"

print(ocr_result_2)
```

```python
# Save the extracted Arabic text to a file
output_filename = 'approach_2.txt'
with open(output_filename, 'w', encoding='utf-8') as f:
    f.write(ocr_result_2)

print(f'Extracted Arabic text saved to {output_filename}')
```

## Pre-Processing

```python
import cv2
import numpy as np
import pytesseract
from PIL import Image

# Load your image with OpenCV
image_cv2 = cv2.imread('page_5.jpg')

if image_cv2 is None:
    print("Error: Could not load image.")
else:
    # Convert BGR (OpenCV default) to RGB for PIL compatibility
    image_rgb = cv2.cvtColor(image_cv2, cv2.COLOR_BGR2RGB)

    # Convert the NumPy array (RGB) to a PIL Image
    pil_image = Image.fromarray(image_rgb)

    # Perform OCR on the PIL image, specifying Arabic language
    ocr_result_3 = pytesseract.image_to_string(pil_image, lang='ara')
    print(ocr_result_3)
```

```python
# Save the extracted Arabic text to a file
output_filename = 'approach_3.txt'
with open(output_filename, 'w', encoding='utf-8') as f:
    f.write(ocr_result_3)

print(f'Extracted Arabic text saved to {output_filename}')
```
