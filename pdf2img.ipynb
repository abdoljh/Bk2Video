```python
import pymupdf
import os

# Ensure clean slate by removing output.pdf if it exists from previous runs
#if os.path.exists('output.pdf'):
    #os.remove('output.pdf')
    #print("Removed residual output.pdf from a previous run.")

# 1. PDF to Image: Convert all pages of chapter2.pdf to individual PNGs
doc = pymupdf.open("chapter2.pdf")
image_files_to_clean_up = [] # Track files to delete later
for page in doc:
    pix = page.get_pixmap(dpi=300) # Higher DPI for better quality
    img_filename = f"page_{page.number}.png"
    pix.save(img_filename)
    image_files_to_clean_up.append(img_filename)
doc.close() # Close the original PDF handler
print(f"Converted {len(image_files_to_clean_up)} pages to individual PNG files.")

# The section that previously generated 'output.pdf' has been removed to ensure it's not created.

# 2. Combine all generated page_*.png files into a single combined_output.pdf
# Use the tracked image_files_to_clean_up list for accuracy, ensuring only generated files are processed.
combined_pdf = pymupdf.open()
images_combined_count = 0
for img_file in image_files_to_clean_up:
    try:
        img_doc_for_combine = pymupdf.open(img_file)
        pdf_bytes_for_combine = img_doc_for_combine.convert_to_pdf()
        temp_pdf = pymupdf.open('pdf', pdf_bytes_for_combine)
        combined_pdf.insert_pdf(temp_pdf)
        img_doc_for_combine.close()
        temp_pdf.close()
        images_combined_count += 1
    except Exception as e:
        print(f"Warning: Could not add {img_file} to combined_output.pdf due to error: {e}")

if images_combined_count > 0:
    combined_pdf.save('combined_output.pdf')
    combined_pdf.close()
    print(f"Successfully combined {images_combined_count} images into combined_output.pdf.")
else:
    print("No images were successfully combined into combined_output.pdf.")


# 3. Delete the generated image files
for img_file in image_files_to_clean_up:
    if os.path.exists(img_file):
        try:
            os.remove(img_file)
            print(f"Deleted intermediate image file: {img_file}")
        except OSError as e:
            print(f"Error deleting {img_file}: {e}")
print("Cleanup of all intermediate image files completed.")
```
