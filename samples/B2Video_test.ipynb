```python
# Cell 02
import sys
import io
from pdfminer.high_level import extract_text

# Provide the path to your PDF file
source_pdf = "Sample6.pdf"
text = extract_text(source_pdf)

print("--- Raw text extracted from PDF ---")
print(text)
print("\n--- Lines after text.splitlines() ---")
lines = text.splitlines()
print(lines)
print("\n--- Processing each line ---")

# Create a string buffer to capture output
output_capture = io.StringIO()
# Redirect stdout to the buffer
original_stdout = sys.stdout
sys.stdout = output_capture

# Iterate through each line and print it along with its hex codes
for line in lines:
    # Skip empty lines to avoid cluttering output with empty entries
    if not line.strip():
        continue

    # Generate hex codes for the original line
    original_hex_codes = []
    for char in line:
        original_hex_codes.append(f"{char} {ord(char):04X}")

    # Reverse the line
    reversed_line = line[::-1]

    # Generate hex codes for the reversed line
    reversed_hex_codes = []
    for char in reversed_line:
        reversed_hex_codes.append(f"{char} {ord(char):04X}")

    print(f"{line}\n {' '.join(original_hex_codes)}\n{reversed_line}\n {' '.join(reversed_hex_codes)}\n")

# Restore stdout to the original (notebook) stdout
sys.stdout = original_stdout

# Get the captured output
captured_output = output_capture.getvalue()

# Print the captured output to the notebook (so it "appears" as before)
print(captured_output)

# Save the captured output to a file
output_filename = source_pdf.replace('.pdf', '.txt')
with open(output_filename, 'w', encoding='utf-8') as f:
    f.write(captured_output)

print(f"Output is saved to '{output_filename}'")
```
